<h5>DATA IMPORTING</h5>

```python
df = pd.read_csv("TWO_CENTURIES_OF_UM_RACES.csv")
```

<h5>DATA SELECTION AND FILTERING</h5>

```python
df[df['Year of event'] == 2020]
```

```python
df[(df['Event distance/length'].isin(['50mi','50km'])) & (df['Year of event'] == 2020)]
```

```python
df[df['Event name'].str.split('(').str.get(1).str.split(')').str.get(0) == 'USA']
```

```python
df1 = df[(df['Event distance/length'].isin(['50mi','50km'])) & (df['Year of event'] == 2020) & (df['Event name'].str.split('(').str.get(1).str.split(')').str.get(0) == 'USA')]
```

```python
df1['Event name'] = df1['Event name'].str.split('(').str.get(0)
```

```python
df1['athlete age'] = 2020 - df1['Athlete year of birth']
```

<h5>DATA TRANSFORMATION</h5>

```python
df2.groupby(['race_distance','athlete_gender'])['avg_speed'].mean()
```

```python
df2.query('race_distance == "50mi"').groupby('athlete_age')['avg_speed'].agg(['mean','count']).sort_values('mean',ascending = True).query('count>9')
```

```python
df2['race_month'] = df2['race_day'].str.split('.').str.get(1).astype(int)
```

```python
#seasons for data -> Slower in summer than winter:

#spring 3-5
#Summer 6-8
#Fall 9-11
#Winter 12-2

#split between two decimals
```

```python
df2.dtypes
```

```python
df2['race_season'] = df2['race_month'].apply(lambda x : 'Winter' if x > 11 else 'Fall' if x > 8 else 'Summer' if x > 5 else 'Spring' if x > 2 else 'Winter')
```

```python
df2
```

```python
df2.groupby('race_season')['avg_speed'].agg(['mean','count']).sort_values('mean',ascending = False)
```

```python
df2.query('race_distance == "50km"').groupby('race_season')['avg_speed'].agg(['mean','count']).sort_values('mean',ascending = False)
```
